# Expfit notes 6: Model selection

## Nested models: likelihood ratio test

Accept a model with $m + 1$ exponential terms over one with $m$ terms if

\begin{align}
\lambda = -2 \left( \ell(\hat\theta_m) - \ell(\hat\theta_{m + 1}) \right) \geq \chi^2_{1,1-\alpha}
\end{align}

Should be chi-square distributed with 1 dof

\begin{align}
\ell(\theta) = -\frac{n}{2}\log(2\pi) - n\log(\sigma) - \frac{n}{2\sigma^2} E(\theta)
\end{align}

On same data, using same $n$ and same (estimate of) $\sigma$

\begin{align}
\lambda = \frac{n}{\sigma^2} \left( E(\hat\theta_{m}) - E(\hat\theta_{m+1}) \right)
\end{align}

Using
\begin{align}
E(\hat\theta_{m}) - E(\hat\theta_{m+1}) < \frac{\sigma^2}{n} \chi^2_{1,1-\alpha}
\end{align}

## Non-nested models

Not as good as likelihood ratio, but can use the [Akaike information criterion](https://en.wikipedia.org/wiki/Akaike_information_criterion):
\begin{align}
\mathcal{A} 
= 2m - 2 \ell(\theta) 
= 2m + n\log(2\pi) + 2n\log(\sigma) + \frac{n}{\sigma^2} \hat{E}
\end{align}
where $m$ is the number of parameters in the model (1 for the offset, and then 2 per exponential).

A lower AIC indicates a better model.
So we select the model with the lowest 
\begin{align}
\mathcal{A}^* = 2m + \frac{n}{\sigma^2} \hat{E}_m
\end{align}
where $\hat{E}_m$ is the best solution found for the $m$-parameter model.

Note that we have to use the same $\sigma^2$ approximation for both models.

### Line versus exponential

Given an exponential fit with three parameters, $E_3$, and a fit of a straight line with two parameters, $E_2$.
Assume the exponential is a better fit: $E_3 < E_2$, then use $E_3 \approx \sigma^2$

Accept line if
\begin{align}
2 \cdot 2 + \frac{n}{\sigma^2} E_2 &\leq 2 \cdot 3 + \frac{n}{\sigma^2} E_3 \\
n \frac{E_2}{E_3} &\leq 2 + n \\
E_2 &\leq \frac{2 + n}{n} E_3
\end{align}

For large $n$ this quickly becomes very small, e.g. $n = 1000 \rightarrow 1.002$

### Line versus exponential, with user-specified noise level

Alternatively, if we have a user-specified noise level:

Accept line if
\begin{align}
E_2 &\leq \frac{2 \sigma^2}{n} + E_3 \\
\end{align}

## Line versus user-specified noise level?

Can we avoid fitting altogether based on $E_2$ and a user-specified $\sigma$ ?

If the fit is perfect, $\sqrt{\hat{E}_2}$ should equal the _sample_ standard deviation, $s$, over the same sample, and approximate the true standard deviation $\sigma$:
\begin{align}
\hat{E}_2 = s^2 \approx \sigma^2
\end{align}
So if $\hat{E}_2$ is close to the user estimate of $\sigma^2$, we can assume it is a straight line.

But how close?

First, $\hat{E}_2$ is derived from linear squares which will likely _overfit_, absorbing some of the variance in the sample and leading to a lower estimate $\hat{E}_2$ than the sample standard deviation estimated from the true parameters.
So if we find $\hat{E}_2$ and the error of the true solution, then $\hat{E}_2 \leq E_{2}(\theta_\text{true})$.

However, if we do not known $E_{2}(\theta_\text{true}$ then we can only estimate it with some $s(\text{sample})$ for some sample that looks particularly flat.
How close will this be to $\hat{E}_2$ or to the true $\sigma$?
One way to answer would be to have a confidence interval around $s$.

For Normally distributed noise, a large-n approximation of the standard deviation in $s$ is $s / \sqrt{2 n}$.
So we could set some upper limit, e.g. accept as line if $E_2 \leq s (1 + 1.96 \sqrt{2 n})$.
But this means we get (A) a new magic number (our confidence level), (B) another equation that depends on normally distributed additive noise, (C) quite far from the goal of fitting an exponential.

So for now we just reject obvious lines, but accept that - when asked to fit an exponential - the routine will sometimes fit an exponential even if that's not the right thing to do.
