# Expfit notes 6: Model selection

## Nested models: likelihood ratio test

Accept a model with $m + 1$ exponential terms over one with $m$ terms if

\begin{align}
\lambda = -2 \left( \ell(\hat\theta_m) - \ell(\hat\theta_{m + 1}) \right) \geq \chi^2_{1,1-\alpha}
\end{align}

Should be chi-square distributed with 1 dof

\begin{align}
\ell(\theta) = -\frac{n}{2}\log(2\pi) - n\log(\sigma) - \frac{n}{2\sigma^2} E(\theta)
\end{align}

On same data, using same $n$ and same (estimate of) $\sigma$

\begin{align}
\lambda = \frac{n}{\sigma^2} \left( E(\hat\theta_{m}) - E(\hat\theta_{m+1}) \right)
\end{align}

Accept the more complex model if the improvement in $E$ (measured as old error minus new error) is above the threshold:
\begin{align}
E(\hat\theta_{m}) - E(\hat\theta_{m+1}) \geq \frac{\sigma^2}{n} \chi^2_{1,1-\alpha}
\end{align}

### Estimating the variance

If the variance is now known, we can fill in $\sigma^2 = E(\hat\theta_{m+1})$ and accept the improved model if

\begin{align}
E(\hat\theta_{m}) \geq \left(1 + \chi^2_{1,1-\alpha} / n\right) E(\hat\theta_{m+1})
\end{align}

or alternatively _reject_ if
\begin{align}
E(\hat\theta_{m}) < \left(1 + \chi^2_{1,1-\alpha} / n\right) E(\hat\theta_{m+1})
\end{align}
or
\begin{align}
\frac{E(\hat\theta_{m})}{E(\hat\theta_{m+1})} < 1 + \chi^2_{1,1-\alpha} / n
\end{align}

Note that $1 + \chi^2_{1,1-\alpha} / n > 1$, so that we will reject correctly even if the new fit does not improve.

## Non-nested models

Not as good as likelihood ratio, but can use the [Akaike information criterion](https://en.wikipedia.org/wiki/Akaike_information_criterion):
\begin{align}
\mathcal{A} 
= 2m - 2 \ell(\theta) 
= 2m + n\log(2\pi) + 2n\log(\sigma) + \frac{n}{\sigma^2} \hat{E}
\end{align}
where $m$ is the number of parameters in the model (1 for the offset, and then 2 per exponential).

A lower AIC indicates a better model.
So we select the model with the lowest 
\begin{align}
\mathcal{A}^* = 2m + \frac{n}{\sigma^2} \hat{E}_m
\end{align}
where $\hat{E}_m$ is the best solution found for the $m$-parameter model.

Note that we have to use the same $\sigma^2$ approximation for both models.

### Line versus exponential

Given an exponential fit with three parameters, $E_3$, and a fit of a straight line with two parameters, $E_2$.
Assume the exponential is a better fit: $E_3 < E_2$, then use $E_3 \approx \sigma^2$

_Accept_ exponential if
\begin{align}
2 \cdot 3 + \frac{n}{\sigma^2} E_3 &< 2 \cdot 2 + \frac{n}{\sigma^2} E_2 \\
2 \frac{\sigma^2}{n} + E_3 &< E_2 \\
E_2 - E_3 &\geq 2 \frac{\sigma^2}{n}
\end{align}

Approximating $\sigma^2 = E_3$, accept if
\begin{align}
\frac{E_2}{E_3} &\geq 1 + \frac{2}{n}
\end{align}
or reject if
\begin{align}
\frac{E_2}{E_3} < 1 + 2 / n
\end{align}

As before, this always rejects if $E_2 < E_3$.
For large $n$, rejection quickly becomes more unlikely, e.g. for $n = 1000$ we accept an improvement of a factor $1.002$.

## Underestimating $\sigma$

The downside to estimating $\sigma$, is that when we overfit, our model absorbs some of the variance in the data and we underestimate $\sigma$.

Above, we accept a higher order exponential if the improvement (old error minus new) is greater than

\begin{align}
\frac{\sigma^2}{n} \chi^2_{1,1-\alpha}
\end{align}

and a single exponential over a line if the improvement exceeds

\begin{align}
\frac{\sigma^2}{n} 2
\end{align}

As a result, underestimating $\sigma$ can lower the threshold for the more complex model to be accepted.
